# Logistic Regression Model

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import matplotlib.pyplot as plt
import seaborn as sns
```

## Data loading and Processing

```python
# LOAD DATA
df = pd.read_csv("prepped_NHIS_dataset.csv")

print("Dataset shape:", df.shape)
print("Risk level distribution:\n", df["Risk_Level"].value_counts())
```

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
df.head()
```

### Feature Assigning and Test/Test Split

```python
# PREPARE FEATURES & TARGET
feature_cols = [col for col in df.columns if col not in ["Data_Value", "Risk_Level"]]

X = df[feature_cols]
y = df["Risk_Level"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y  # IMPORTANT for class balance
)
```

```python
feature_cols
```

## Training and Fitting Data into the Logistic Regression Pipeline

```python
logreg_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression(
        multi_class="multinomial",
        solver="lbfgs",
        max_iter=1000,
        n_jobs=-1
    ))
])

logreg_pipeline.fit(X_train, y_train)
```

```python
y_train_pred = logreg_pipeline.predict(X_train)
y_test_pred = logreg_pipeline.predict(X_test)
```

```python
train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)
```

## Evaluation Metrics

```python
# Performance Metrics

print(f"Training Accuracy: {train_acc:.4f}")
print(f"Test Accuracy:     {test_acc:.4f}\n")

print("Classification Report (Test):")
print(classification_report(y_test, y_test_pred))
```

61% Accuracy, so lets try to up those numbers.

```python
logreg_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression(
        multi_class="multinomial",
        solver="lbfgs",
        max_iter=3000,
        n_jobs=-1
    ))
])

logreg_pipeline.fit(X_train, y_train)
```

```python
y_train_pred = logreg_pipeline.predict(X_train)
y_test_pred = logreg_pipeline.predict(X_test)
```

```python
train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)
```

```python
# Performance Metrics

print(f"Training Accuracy: {train_acc:.4f}")
print(f"Test Accuracy:     {test_acc:.4f}\n")

print("Classification Report (Test):")
print(classification_report(y_test, y_test_pred))
```

### Define soft class weight candidates

```python
class_weight_grid = [
    {0:1, 1:1,   2:1,   3:1},      # baseline
    {0:1, 1:1.1, 2:1.2, 3:1.4},
    {0:1, 1:1.2, 2:1.4, 3:1.8},
    {0:1, 1:1.3, 2:1.6, 3:2.2},
    {0:1, 1:1.5, 2:2.0, 3:3.0},    # close to balanced
]
```

```python
param_grid = {
    "logreg__C": [0.1, 0.3, 1, 3],
    "logreg__class_weight": class_weight_grid
}

grid = GridSearchCV(
    logreg_pipeline,
    param_grid,
    scoring="balanced_accuracy",
    cv=5,
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Best params:", grid.best_params_)
print("Best CV balanced accuracy:", grid.best_score_)
```

### Evaluation with Tuned Hyperparameter

```python
best_pipeline = grid.best_estimator_

y_test_pred = best_pipeline.predict(X_test)

print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
print(classification_report(y_test, y_test_pred))
```

The model has almost ignored the class "3" which is the high risk class.

Going to merge the 2 and 3 classes so we will have:

* 0 - Low risk

* 1 - Medium risk

* 2 - High risk

* 3 - High Risk (Merged with 2)

Initial experiments with four risk categories showed severe confusion between higher risk levels, particularly Risk_Level 2 and 3. As these categories represent similar clinical severity, they were merged into a single high-risk class, resulting in improved class balance and more reliable predictive performance.

```python
df["Risk_Level_3"] = df["Risk_Level"].replace({3: 2})

print(df["Risk_Level_3"].value_counts())
```

```python
X = df[feature_cols]
y = df["Risk_Level_3"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
```

```python
print(y_train.unique())
print(y_test.unique())
```

```python
class_weight = {
    0: 1,
    1: 1.2,
    2: 1.6
}
```

```python
logreg_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression(
        multi_class="multinomial",
        solver="lbfgs",
        C=0.01,
        class_weight=class_weight,
        max_iter=4000,
        n_jobs=-1
    ))
])

logreg_pipeline.fit(X_train, y_train)
```

```python
y_pred = logreg_pipeline.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
class_weight_grid = [
    {0:1, 1:1,   2:1},     # no weighting
    {0:1, 1:1.1, 2:1.3},
    {0:1, 1:1.2, 2:1.6},
    {0:1, 1:1.3, 2:1.8},
    {0:1, 1:1.5, 2:2.0}
]
```

```python
param_grid = {
    "logreg__C": [0.1, 0.3, 1, 3, 5, 10],
    "logreg__class_weight": class_weight_grid
}

grid = GridSearchCV(
    logreg_pipeline,
    param_grid=param_grid,
    scoring="balanced_accuracy",
    cv=5,
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Best C:", grid.best_params_)
print("CV Accuracy:", grid.best_score_)
```

```python
logreg_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression(
        multi_class="multinomial",
        solver="lbfgs",
        C=3,
        class_weight = {0: 1, 1: 1.1, 2: 1.3},
        n_jobs=-1
    ))
])

logreg_pipeline.fit(X_train, y_train)
```

```python
y_pred = logreg_pipeline.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

Although Logistic Regression was carefully tuned using class weighting, regularization, and threshold adjustment, performance gains plateaued, particularly for higher risk classes. The inability of threshold calibration to improve recall without degrading overall performance suggests limited separability within a linear decision framework. As a result, Logistic Regression was considered to have reached its practical performance ceiling for this task, motivating the exploration of more expressive non-linear models
